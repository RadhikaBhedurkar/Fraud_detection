```python
# Fraud vs Non-Fraud Distribution
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Class', hue='Class', data=df, palette='coolwarm', legend=False)
plt.title('Distribution of Legitimate vs Fraudulent Transactions')
plt.xlabel('Transaction Type (0 = Legit, 1 = Fraud)')
plt.ylabel('Count')
plt.show()
```

# Insight:
~99.83% of transactions are legitimate; only ~0.17% are fraudulent.

```python
# Transaction Amount Analysis
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x='Class', y='Amount', hue='Class', data=df, palette='husl', legend=False)
plt.yscale('log')
plt.title('Transaction Amount Distribution by Class')
plt.xlabel('Transaction Type (0 = Legit, 1 = Fraud)')
plt.ylabel('Transaction Amount ($)')
plt.show()
```

# Insight:
Fraudulent transactions tend to have lower amounts, but a few high-value ones exist.

```python
# Time-based Fraud Patterns
```

```python
plt.figure(figsize=(8,4))
sns.histplot(df[df['Class']==0]['Time'], bins=50, color='green', label='Legit', alpha=0.6)
sns.histplot(df[df['Class']==1]['Time'], bins=50, color='red', label='Fraud', alpha=0.6)
plt.legend()
plt.title('Transaction Time Distribution')
plt.xlabel('Time (seconds)')
plt.ylabel('Frequency')
plt.show()
```

# Insight:
Frauds might cluster at specific time ranges (e.g., off-peak hours).

```python
# Correlation Heatmap
```

```python
plt.figure(figsize=(10,5))
corr = df.corr()
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Feature Correlation Heatmap')
plt.show()
```

# Insight:
Certain features (like V17, V14, V12) have strong negative or positive correlation with the Class (fraud).
